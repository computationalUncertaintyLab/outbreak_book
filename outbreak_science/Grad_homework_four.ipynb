## What exactly is the Student T distribution (a multi-part journey)? 

If our goal is to characterize the total number of excess deaths due to influenza and/or pnemonia in the states that implemented an early vs late public health response then we can use the Central Limit Theorem (CLT). 

The CLT says that if we collect $n$ data points $(d_{1},d_{2},\cdots,d_{n})$, and we assume that these data points were generated from $n$ random variables $(X_{1}, X_{2}, \cdots, X_{n})$, and we assume that every random variable ${X_{i}}_{i=1}^{N}$ all have the same distribution, and we assume that these random variables are independent (sheesh) then 
$ \frac{\sum_{i=1}^{n} X_{i}}{n} \sim \mathcal{N}(\mu, \frac{\sigma^{2}}{n} )$

Our goal is to learn about $\mu$ and we assume that this parameter is unknown. 
We need to figure out $\sigma^{2}$ but we have a problem. We don't know $\sigma^{2}$. We need to estimate it.

## The Chi-squared distribution and how its related to the variance (part 1)

Given a standard normal density $Z\sim \mathcal{N}(0,1)$ the random variable that is geenrated by squaring $Z$ is called a Chi-squared distribution with one degree of freedom. 
We can write this mathematically as $Y = Z^{2} \sim \chi^{2}_{1}$. 
Even more, the sum of $n$ squared normal densities is called a Chi-squared distribution with $n$ degrees of freedom. 
\begin{align}
    \sum_{i=1}^{n} Y_{i} \sim \chi^{2}_{n} ; \text{ where } \; Y_{i}  \sim \mathcal{N}(0,1)
\end{align}

## From variance to Chi squared (Part 2)

If we assume that our data $\mathcal{D} = (x_{1}, x_{2}, \cdots, x_{n})$ are all normally distributed $(x_{i} \sim \mathcal{N}(\mu,\sigma^{2}) )$ then what does the distribution of our variance even look like? 
In other words, what does the probability distribution of this thing look like
\begin{align}
    S^{2} = \frac{1}{n-1} \sum_{ i=1 }^{n} (X_{i} - \bar{X})^{2}
\end{align}
We know that the variance has to be positive and we know (by **assumption**) that all of the individual data points were generated as $X_{i} \sim \mathcal{N}(\mu,\sigmas^{2})$.  
Well, there are some super useful properties of the Normal density that are worth reviewing related to translation (i.e. shifting) and scaling. 

Here they are:
Given $X \sim \mathcal{N}(\mu,\sigma^{2})$
\begin{align}
    Y = X +c &\implies Y \sim \mathcal{N}(\mu+c, \sigma^{2}) \\ 
    Y = cX   &\implies Y \sim \mathcal{N}(\mu, c^{2}\sigma^{2}) \\ 
\end{align}

Then this means that we can take one of our data points and translate/scale it so that it turns into a standard normal density. 

\begin{align}
    X_{i} \sim \mathcal{N}(\mu, \sigma^{2}) \\ 
    \frac{X_{i} - \mu}{\sqrt{\sigma^{2}}} \sim \mathcal{N}(0,1) \\ 
\end{align}

Even more, if we square this then we have a chi squared density. Nice. 

\begin{align}
    \frac{\left(X_{i} - \mu\right)^{2}}{\sigma^{2}} \sim \chi_{1}^{2} \\ 
\end{align}

Ok but this is only a single data point. 
How should we handle $n$ data points? 
First things first, we want to get closer to understanding the distribution of $S$ and $S$ has $\bar{X}$ in it. Then lets add it.

\begin{align}
    \frac{\left(X_{i} - \bar{X} + \bar{X} - \mu\right)^{2}}{\sigma^{2}} \sim \chi_{1}^{2} \\ 
\end{align}

We can expand that squared term treating $X_{i} - \bar{X}$ and $\bar{X} - \mu$ as the two terms in our squared expression. 

\begin{align}
\frac{\left(X_{i} - \bar{X}\right)^{2} + \left(\bar{X} - \mu\right)^{2} - 2\left(X_{i} - \bar{X}\right)\left(\bar{X} - \mu\right) }{\sigma^{2}} \sim \chi_{1}^{2} \\ 
\end{align}

This looks promising because that first term, $\frac{\left(X_{i} - \bar{X}\right)^{2}}{\sigma^{2}}$ is pretty close to $S$. To get closer we need to sum over all $n$ data points. 

\begin{align}
    \sum_{i=1}^{n} \frac{\left(X_{i} - \bar{X}\right)^{2} + \left(\bar{X} - \mu\right)^{2} -2\left(X_{i} - \bar{X}\right)\left(\bar{X} - \mu\right)}{\sigma^{2}} \sim \chi_{n}^{2} \\ 
    \sum_{i=1}^{n} \frac{\left(X_{i} - \bar{X}\right)^{2}}{\sigma^{2}} + n\frac{\left(\bar{X} - \mu\right)^{2}}{\sigma^{2}} \\
    \sum_{i=1}^{n} \frac{\left(X_{i} - \bar{X}\right)^{2}}{\sigma^{2}} + \frac{\left(\bar{X} - \mu\right)^{2}}{\sigma^{2}/n}  \sim \chi_{n}^{2}
\end{align}

Well what is the distribution for this
\begin{align}
    \frac{\left(\bar{X} - \mu\right)^{2}}{\sigma^{2}/n} 
\end{align}

The variance for $\bar{X}$ is $\sigma^{2}/n$ and from the CLT $\bar{X} \sim \mathcal{N}(\mu,\sigma^{2}/n)$. That means then that 

\begin{align}
    \frac{\left(\bar{X} - \mu\right)^{2}}{\sigma^{2}/n} \sim \chi_{1}^{2}
\end{align}

so then 

\begin{align}
\sum_{i=1}^{n} \frac{\left(X_{i} - \bar{X}\right)^{2}}{\sigma^{2}} + \chi_{1}^{2} \sim \chi_{n}^{2} \\ 
\end{align}

and 

\begin{align}
\sum_{i=1}^{n} \frac{\left(X_{i} - \bar{X}\right)^{2}}{\sigma^{2}}  \sim \chi_{n-1}^{2} 
\end{align}

Then we can substitue in

\begin{align}
\frac{(n-1) S^{2}}{\sigma^{2}}  \sim \chi_{n-1}^{2} 
\end{align}

Lets use this information to compute a 95% confidence interval for our estimate of $\sigma^{2}$ in the early and late public health action states.

## A t-test for public health response timing and disease burden for the 1918 flu pandemic

A two sample t-test provides evidence for (or against) the hypothesis 

\begin{align}
    H_{\text{null}}&: \mu_{\text{early}} - \mu_{\text{late}} = 0 \\  
    H_{\text{alternative}}&:\mu_{\text{early}} - \mu_{\text{late}} \neq 0 \\  
\end{align}

When we don't know the variance from our sample (per usual) 

\begin{align}
    T = \frac{ ( \bar{X}_{\text{early}} - \bar{X}_{\text{late}})  - 0 }{ SE }\\
    SE = \sqrt{\frac{S_{\text{early}}^{2}}{n_{\text{early}}}  + \frac{S_{\text{late}}^{2}}{n_{\text{late}}} }
\end{align}

Because we are estimating the variances, this statistic has a T distribution with (assuming the null hypothesis is true) expected value 0 or $\mathcal{T}(0,SE,df)$ .
The degrees of freedom are beyond the scope of this homework. Note that they are usually approximated.
